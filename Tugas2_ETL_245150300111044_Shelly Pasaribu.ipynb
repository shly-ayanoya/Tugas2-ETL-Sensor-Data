```python
# ============================================================
# COMBINE DATA MENJADI SATU DATAFRAME DENGAN SCHEMA KONSISTEN
# ============================================================

standard_columns = [
    "timestamp",
    "sensor_id",
    "location",
    "temperature_celsius",
    "humidity_percent",
    "pressure_hpa",
    "air_quality_aqi",
    "status"
]

# ------------------------------------------------------------
# A. Membaca CSV utama
# ------------------------------------------------------------

main_csv = data_path / "sensor_data_main.csv"

try:
    df_main = pd.read_csv(main_csv)
    print("sensor_data_main.csv berhasil dibaca.")
except FileNotFoundError:
    print(f"File tidak ditemukan: {main_csv}")


# ------------------------------------------------------------
# B. Membaca CSV dengan separator ;
# ------------------------------------------------------------

semicolon_csv = data_path / "sensor_data_semicolon.csv"

try:
    df_semicolon = pd.read_csv(
        semicolon_csv,
        sep=";"
    )
    print("sensor_data_semicolon.csv berhasil dibaca.")
except FileNotFoundError:
    print(f"File tidak ditemukan: {semicolon_csv}")


# ------------------------------------------------------------
# C. Membaca seluruh sheet Excel
# ------------------------------------------------------------

excel_file = data_path / "sensor_data_monthly.xlsx"

try:
    excel_sheets = pd.read_excel(
        excel_file,
        sheet_name=None
    )

    print("sensor_data_monthly.xlsx berhasil dibaca.")
    print("Sheet yang ditemukan:", list(excel_sheets.keys()))

except FileNotFoundError:
    print(f"File tidak ditemukan: {excel_file}")


# ------------------------------------------------------------
# D. Menstandarkan schema CSV utama
# ------------------------------------------------------------

df_main_standard = df_main.copy()

df_main_standard.columns = [
    col.strip().lower()
    for col in df_main_standard.columns
]

df_main_standard = df_main_standard[standard_columns]


# ------------------------------------------------------------
# E. Menstandarkan schema CSV separator ;
# ------------------------------------------------------------

df_semicolon_standard = df_semicolon.copy()

df_semicolon_standard.columns = [
    col.strip().lower()
    for col in df_semicolon_standard.columns
]

df_semicolon_standard = df_semicolon_standard[standard_columns]


# ------------------------------------------------------------
# F. Menstandarkan dan menggabungkan seluruh sheet Excel
# ------------------------------------------------------------

excel_dataframes = []

for sheet_name, df_sheet in excel_sheets.items():

    df_sheet_standard = df_sheet.copy()

    df_sheet_standard.columns = [
        col.strip().lower()
        for col in df_sheet_standard.columns
    ]

    df_sheet_standard = df_sheet_standard[standard_columns]

    excel_dataframes.append(df_sheet_standard)


df_excel_standard = pd.concat(
    excel_dataframes,
    ignore_index=True
)


# ------------------------------------------------------------
# G. Menggabungkan seluruh sumber data
# ------------------------------------------------------------

df_combined_raw = pd.concat(
    [
        df_main_standard,
        df_semicolon_standard,
        df_excel_standard
    ],
    ignore_index=True
)


# ------------------------------------------------------------
# H. Menghapus data yang identik
# ------------------------------------------------------------

df_combined = df_combined_raw.drop_duplicates(
    subset=standard_columns
).reset_index(drop=True)


# ------------------------------------------------------------
# I. Memastikan schema akhir konsisten
# ------------------------------------------------------------

df_combined = df_combined[standard_columns]

print("\n=== HASIL COMBINE DATA ===")
print(f"Data CSV utama       : {len(df_main_standard)} baris")
print(f"Data CSV separator ; : {len(df_semicolon_standard)} baris")
print(f"Data Excel           : {len(df_excel_standard)} baris")
print(f"Data sebelum deduplikasi : {len(df_combined_raw)} baris")
print(f"Data setelah deduplikasi: {len(df_combined)} baris")

print("\n=== SCHEMA AKHIR ===")
print(df_combined.columns.tolist())

print("\n=== CONTOH DATA ===")
display(df_combined.head())
```

```python
# ============================================================
# NO. 2 BAGIAN 1 - DATA CLEANING
# Hapus duplikat, standarisasi nama kolom,
# dan konversi tipe data
# ============================================================

# Membuat salinan data hasil combine
df_clean = df_combined.copy()


# ------------------------------------------------------------
# 1. HAPUS DUPLIKAT
# ------------------------------------------------------------

print("=== SEBELUM MENGHAPUS DUPLIKAT ===")
print(f"Jumlah baris : {len(df_clean)}")
print(f"Jumlah duplikat : {df_clean.duplicated().sum()}")

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\n=== SETELAH MENGHAPUS DUPLIKAT ===")
print(f"Jumlah baris : {len(df_clean)}")
print(f"Jumlah duplikat : {df_clean.duplicated().sum()}")


# ------------------------------------------------------------
# 2. STANDARISASI NAMA KOLOM
# ------------------------------------------------------------

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\n=== NAMA KOLOM SETELAH STANDARISASI ===")
print(df_clean.columns.tolist())


# ------------------------------------------------------------
# 3. KONVERSI TIPE DATA
# ------------------------------------------------------------

# Timestamp menjadi datetime
df_clean["timestamp"] = pd.to_datetime(
    df_clean["timestamp"],
    errors="coerce"
)

# Kolom numerik
numeric_columns = [
    "temperature_celsius",
    "humidity_percent",
    "pressure_hpa",
    "air_quality_aqi"
]

for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column],
        errors="coerce"
    )

# Kolom kategorikal
categorical_columns = [
    "sensor_id",
    "location",
    "status"
]

for column in categorical_columns:
    df_clean[column] = df_clean[column].astype("category")


# ------------------------------------------------------------
# 4. CEK HASIL KONVERSI TIPE DATA
# ------------------------------------------------------------

print("\n=== TIPE DATA SETELAH KONVERSI ===")
print(df_clean.dtypes)


# ------------------------------------------------------------
# 5. CEK HASIL DATA CLEANING
# ------------------------------------------------------------

print("\n=== HASIL DATA CLEANING BAGIAN 1 ===")
print(f"Jumlah baris : {df_clean.shape[0]}")
print(f"Jumlah kolom : {df_clean.shape[1]}")

print("\n=== MISSING VALUE SETELAH KONVERSI ===")
print(df_clean.isnull().sum())

print("\n=== CONTOH DATA ===")
display(df_clean.head())
```

```python
# ============================================================
# NO. 2 BAGIAN 2 - NORMALISASI NILAI KATEGORIKAL
# ============================================================

# ------------------------------------------------------------
# 1. Melihat nilai kategorikal sebelum normalisasi
# ------------------------------------------------------------

print("=== NILAI LOCATION SEBELUM NORMALISASI ===")
print(df_clean["location"].value_counts(dropna=False))

# ------------------------------------------------------------
# 2. Normalisasi kolom location
# ------------------------------------------------------------

df_clean["location"] = (
    df_clean["location"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# ------------------------------------------------------------
# 3. Normalisasi kolom status
# ------------------------------------------------------------

df_clean["status"] = (
    df_clean["status"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# ------------------------------------------------------------
# 4. Kembalikan tipe location dan status menjadi category
# ------------------------------------------------------------

df_clean["location"] = df_clean["location"].astype("category")
df_clean["status"] = df_clean["status"].astype("category")

# ------------------------------------------------------------
# 5. Melihat nilai setelah normalisasi
# ------------------------------------------------------------

print("\n=== NILAI LOCATION SETELAH NORMALISASI ===")
print(df_clean["location"].value_counts(dropna=False))

print("\n=== NILAI STATUS SETELAH NORMALISASI ===")
print(df_clean["status"].value_counts(dropna=False))

# ------------------------------------------------------------
# 6. Mengecek jumlah kategori unik
# ------------------------------------------------------------

print("\n=== KATEGORI UNIK ===")

print("Location:")
print(df_clean["location"].cat.categories.tolist())

print("\nStatus:")
print(df_clean["status"].cat.categories.tolist())

# ------------------------------------------------------------
# 7. Menampilkan contoh data setelah normalisasi
# ------------------------------------------------------------

print("\n=== CONTOH DATA SETELAH NORMALISASI ===")
display(
    df_clean[
        ["sensor_id", "location", "status"]
    ].head(10)
)
```

```python
# ============================================================
# NO. 2 BAGIAN 3 - DETEKSI OUTLIER IQR DAN Z-SCORE
# ============================================================
import numpy as np
from scipy.stats import zscore

numeric_columns = [
    "temperature_celsius",
    "humidity_percent",
    "pressure_hpa",
    "air_quality_aqi"
]

# Deteksi outlier menggunakan IQR
iqr_results = {}
for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outlier_mask = (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    )
    iqr_results[column] = {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outlier_mask.sum()
    }

print("=== HASIL DETEKSI OUTLIER IQR ===")
for column, result in iqr_results.items():
    print(f"\n{column}")
    print(f"Q1            : {result['Q1']:.2f}")
    print(f"Q3            : {result['Q3']:.2f}")
    print(f"IQR           : {result['IQR']:.2f}")
    print(f"Batas bawah   : {result['lower_bound']:.2f}")
    print(f"Batas atas    : {result['upper_bound']:.2f}")
    print(f"Jumlah outlier: {result['outlier_count']}")

# Deteksi outlier menggunakan Z-score
zscore_results = {}
for column in numeric_columns:
    valid_data = df_clean[column].dropna()
    z_scores = pd.Series(
        zscore(valid_data),
        index=valid_data.index
    )
    outlier_mask = z_scores.abs() > 3
    zscore_results[column] = {
        "outlier_count": outlier_mask.sum(),
        "mean": valid_data.mean(),
        "std": valid_data.std()
    }

print("\n=== HASIL DETEKSI OUTLIER Z-SCORE ===")
for column, result in zscore_results.items():
    print(f"\n{column}")
    print(f"Mean          : {result['mean']:.2f}")
    print(f"Std           : {result['std']:.2f}")
    print(f"Jumlah outlier: {result['outlier_count']}")

# Perbandingan IQR dan Z-score
comparison = pd.DataFrame({
    "IQR": [iqr_results[column]["outlier_count"] for column in numeric_columns],
    "Z-score": [zscore_results[column]["outlier_count"] for column in numeric_columns]
}, index=numeric_columns)

print("\n=== PERBANDINGAN IQR DAN Z-SCORE ===")
display(comparison)

# Membuat mask outlier dari kedua metode
iqr_mask_all = pd.Series(False, index=df_clean.index)
zscore_mask_all = pd.Series(False, index=df_clean.index)

for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    iqr_mask_all |= (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    )
    valid_data = df_clean[column].dropna()
    z_scores = pd.Series(zscore(valid_data), index=valid_data.index)
    zscore_mask = z_scores.abs() > 3
    zscore_mask_all.loc[zscore_mask.index] |= zscore_mask

print("\n=== RINGKASAN OUTLIER ===")
print(f"Baris terdeteksi IQR      : {iqr_mask_all.sum()}")
print(f"Baris terdeteksi Z-score  : {zscore_mask_all.sum()}")
print(f"Baris terdeteksi keduanya : {(iqr_mask_all & zscore_mask_all).sum()}")
```

```python
# ============================================================
# STRATEGI 1 - FORWARD FILL
# ============================================================

df_ffill = df_clean.copy()

# Urutkan berdasarkan sensor dan waktu
df_ffill = df_ffill.sort_values(
    ["sensor_id", "timestamp"]
).reset_index(drop=True)

# Kolom numerik yang memiliki kemungkinan missing
numeric_columns = [
    "temperature_celsius",
    "humidity_percent",
    "pressure_hpa",
    "air_quality_aqi"
]

# Forward fill berdasarkan masing-masing sensor
df_ffill[numeric_columns] = (
    df_ffill.groupby("sensor_id", observed=True)[numeric_columns]
    .ffill()
)

print("=== MISSING VALUE SETELAH FORWARD FILL ===")
print(df_ffill[numeric_columns].isnull().sum())
print(f"\nTotal missing value: {df_ffill[numeric_columns].isnull().sum().sum()}")
```

```python
# ============================================================
# STRATEGI 2 - INTERPOLASI LINEAR
# ============================================================

df_interpolated = df_clean.copy()

df_interpolated = df_interpolated.sort_values(
    ["sensor_id", "timestamp"]
).reset_index(drop=True)

df_interpolated[numeric_columns] = (
    df_interpolated.groupby("sensor_id", observed=True)[numeric_columns]
    .transform(lambda group: group.interpolate(method="linear"))
)

print("=== MISSING VALUE SETELAH INTERPOLASI LINEAR ===")
print(df_interpolated[numeric_columns].isnull().sum())
print(f"\nTotal missing value: {df_interpolated[numeric_columns].isnull().sum().sum()}")
```

```python
# ============================================================
# PERBANDINGAN HASIL IMPUTASI
# ============================================================

comparison_missing = pd.DataFrame({
    "Sebelum": df_clean[numeric_columns].isnull().sum(),
    "Forward Fill": df_ffill[numeric_columns].isnull().sum(),
    "Interpolasi Linear": df_interpolated[numeric_columns].isnull().sum()
})

print("=== PERBANDINGAN MISSING VALUE ===")
display(comparison_missing)

print("\n=== TOTAL MISSING VALUE ===")
print("Sebelum            :", df_clean[numeric_columns].isnull().sum().sum())
print("Forward Fill       :", df_ffill[numeric_columns].isnull().sum().sum())
print("Interpolasi Linear :", df_interpolated[numeric_columns].isnull().sum().sum())
```

```python
# ============================================================
# NO. 4 BAGIAN 1 - FITUR WAKTU
# ============================================================

# Pastikan timestamp bertipe datetime
df_clean["timestamp"] = pd.to_datetime(df_clean["timestamp"])

# Ekstraksi fitur waktu
df_clean["hour"] = df_clean["timestamp"].dt.hour
df_clean["day"] = df_clean["timestamp"].dt.day
df_clean["week"] = df_clean["timestamp"].dt.isocalendar().week.astype(int)

# Kategori periode waktu
def get_time_period(hour):
    if 5 <= hour < 11:
        return "pagi"
    elif 11 <= hour < 15:
        return "siang"
    elif 15 <= hour < 18:
        return "sore"
    else:
        return "malam"

df_clean["time_period"] = df_clean["hour"].apply(get_time_period)

print("=== FITUR WAKTU ===")
display(
    df_clean[
        ["timestamp", "hour", "day", "week", "time_period"]
    ].head(10)
)

print("\n=== DISTRIBUSI PERIODE WAKTU ===")
print(df_clean["time_period"].value_counts())
```

```python
# ============================================================
# NO. 4 BAGIAN 2 - KONVERSI SUHU, AQI, DAN HEAT INDEX
# ============================================================

# Konversi suhu
df_clean["temperature_fahrenheit"] = (
    df_clean["temperature_celsius"] * 9 / 5
) + 32

df_clean["temperature_kelvin"] = (
    df_clean["temperature_celsius"] + 273.15
)

# Kategori AQI
def categorize_aqi(aqi):
    if pd.isna(aqi):
        return "tidak diketahui"
    elif aqi <= 50:
        return "baik"
    elif aqi <= 100:
        return "sedang"
    elif aqi <= 150:
        return "tidak sehat bagi kelompok sensitif"
    elif aqi <= 200:
        return "tidak sehat"
    elif aqi <= 300:
        return "sangat tidak sehat"
    else:
        return "berbahaya"

df_clean["aqi_category"] = df_clean["air_quality_aqi"].apply(categorize_aqi)

# Heat Index menggunakan suhu Fahrenheit dan kelembapan relatif
T = df_clean["temperature_fahrenheit"]
RH = df_clean["humidity_percent"]

heat_index = (
    -42.379
    + 2.04901523 * T
    + 10.14333127 * RH
    - 0.22475541 * T * RH
    - 0.00683783 * T**2
    - 0.05481717 * RH**2
    + 0.00122874 * T**2 * RH
    + 0.00085282 * T * RH**2
    - 0.00000199 * T**2 * RH**2
)

# Heat Index hanya dihitung pada kondisi yang sesuai
df_clean["heat_index_fahrenheit"] = heat_index.where(
    (T >= 80) & (RH >= 40)
)

print("=== HASIL FEATURE ENGINEERING ===")
display(
    df_clean[
        [
            "temperature_celsius",
            "temperature_fahrenheit",
            "temperature_kelvin",
            "humidity_percent",
            "air_quality_aqi",
            "aqi_category",
            "heat_index_fahrenheit"
        ]
    ].head(10)
)

print("\n=== DISTRIBUSI KATEGORI AQI ===")
print(df_clean["aqi_category"].value_counts())
```

```python
# ============================================================
# NO. 4 BAGIAN 3 - ROLLING STATISTICS 24 JAM
# ============================================================

# Urutkan data berdasarkan sensor dan waktu
df_clean = df_clean.sort_values(
    ["sensor_id", "timestamp"]
).reset_index(drop=True)

# Moving average suhu 24 jam per sensor
df_clean["temperature_ma_24h"] = (
    df_clean.groupby("sensor_id", observed=True)["temperature_celsius"]
    .transform(lambda x: x.rolling(window=24, min_periods=1).mean())
)

# Moving average kelembapan 24 jam per sensor
df_clean["humidity_ma_24h"] = (
    df_clean.groupby("sensor_id", observed=True)["humidity_percent"]
    .transform(lambda x: x.rolling(window=24, min_periods=1).mean())
)

print("=== ROLLING STATISTICS 24 JAM ===")
display(
    df_clean[
        [
            "timestamp",
            "sensor_id",
            "temperature_celsius",
            "temperature_ma_24h",
            "humidity_percent",
            "humidity_ma_24h"
        ]
    ].head(30)
)
```

```python
# ============================================================
# NO. 4 BAGIAN 4 - FITUR ORISINAL
# ============================================================

# Selisih suhu saat ini dengan rata-rata suhu 24 jam
df_clean["temperature_deviation_24h"] = (
    df_clean["temperature_celsius"] -
    df_clean["temperature_ma_24h"]
)

print("=== FITUR ORISINAL: TEMPERATURE DEVIATION 24 JAM ===")
display(
    df_clean[
        [
            "timestamp",
            "sensor_id",
            "temperature_celsius",
            "temperature_ma_24h",
            "temperature_deviation_24h"
        ]
    ].head(20)
)
```

```python
# ============================================================
# NO. 5 BAGIAN 1 - DATA VALIDATION
# ============================================================

print("=== VALIDASI DATA HASIL ETL ===")

# Completeness
total_cells = df_clean.shape[0] * df_clean.shape[1]
missing_cells = df_clean.isnull().sum().sum()
completeness = (1 - missing_cells / total_cells) * 100

# Uniqueness
duplicate_rows = df_clean.duplicated().sum()
uniqueness = (1 - duplicate_rows / len(df_clean)) * 100

# Validity berdasarkan range sensor
valid_temperature = df_clean["temperature_celsius"].between(-10, 50).mean() * 100
valid_humidity = df_clean["humidity_percent"].between(0, 100).mean() * 100
valid_pressure = df_clean["pressure_hpa"].between(950, 1050).mean() * 100
valid_aqi = df_clean["air_quality_aqi"].between(0, 500).mean() * 100

validity = (
    valid_temperature +
    valid_humidity +
    valid_pressure +
    valid_aqi
) / 4

# Consistency
valid_timestamp = df_clean["timestamp"].notna().mean() * 100
valid_sensor_id = df_clean["sensor_id"].notna().mean() * 100
valid_location = df_clean["location"].notna().mean() * 100
consistency = (
    valid_timestamp +
    valid_sensor_id +
    valid_location
) / 3

validation_result = pd.DataFrame({
    "Quality Metric": [
        "Completeness",
        "Uniqueness",
        "Validity",
        "Consistency"
    ],
    "Score (%)": [
        completeness,
        uniqueness,
        validity,
        consistency
    ]
})

display(validation_result)

print("\n=== VALIDASI RANGE SENSOR ===")
print(f"Temperature valid : {valid_temperature:.2f}%")
print(f"Humidity valid    : {valid_humidity:.2f}%")
print(f"Pressure valid    : {valid_pressure:.2f}%")
print(f"AQI valid         : {valid_aqi:.2f}%")
```

```python
# ============================================================
# PERBANDINGAN KUALITAS SEBELUM DAN SESUDAH ETL
# ============================================================

before_completeness = (
    1 - df_combined.isnull().sum().sum() /
    (df_combined.shape[0] * df_combined.shape[1])
) * 100

before_uniqueness = (
    1 - df_combined.duplicated().sum() /
    len(df_combined)
) * 100
after_completeness = completeness
after_uniqueness = uniqueness

quality_comparison = pd.DataFrame({
    "Metric": [
        "Completeness",
        "Uniqueness"
    ],
    "Before ETL (%)": [
        before_completeness,
        before_uniqueness
    ],
    "After ETL (%)": [
        after_completeness,
        after_uniqueness
    ]
})

display(quality_comparison)
```

```python
# ============================================================
# NO. 5 BAGIAN 3 - LOAD KE CSV
# ============================================================

output_path = Path("data/processed")
output_path.mkdir(parents=True, exist_ok=True)

csv_file = output_path / "sensor_data_clean.csv"

df_clean.to_csv(csv_file, index=False)

print(f"CSV berhasil disimpan: {csv_file}")
print(f"Ukuran file: {csv_file.stat().st_size:,} bytes")
```

```python
# ============================================================
# NO. 5 BAGIAN 4 - LOAD KE EXCEL MULTI-SHEET
# ============================================================

location_summary = (
    df_clean.groupby("location", observed=True)
    .agg(
        avg_temperature=("temperature_celsius", "mean"),
        avg_humidity=("humidity_percent", "mean"),
        avg_pressure=("pressure_hpa", "mean"),
        avg_aqi=("air_quality_aqi", "mean"),
        total_records=("sensor_id", "count")
    )
    .reset_index()
)

sensor_summary = (
    df_clean.groupby("sensor_id", observed=True)
    .agg(
        avg_temperature=("temperature_celsius", "mean"),
        avg_humidity=("humidity_percent", "mean"),
        avg_pressure=("pressure_hpa", "mean"),
        avg_aqi=("air_quality_aqi", "mean"),
        total_records=("sensor_id", "count")
    )
    .reset_index()
)

excel_output = output_path / "sensor_data_clean.xlsx"

with pd.ExcelWriter(excel_output, engine="openpyxl") as writer:
    df_clean.to_excel(writer, sheet_name="Clean Data", index=False)
    location_summary.to_excel(writer, sheet_name="By Location", index=False)
    sensor_summary.to_excel(writer, sheet_name="By Sensor", index=False)

print(f"Excel berhasil disimpan: {excel_output}")
print(f"Ukuran file: {excel_output.stat().st_size:,} bytes")
```

```python
print("=== AGGREGATE PER LOCATION ===")
display(location_summary)

print("=== AGGREGATE PER SENSOR ===")
display(sensor_summary)
```

```python
# ============================================================
# NO. 5 BAGIAN 5 - LOAD KE JSON
# ============================================================

json_output = output_path / "sensor_data_clean.json"

df_clean.to_json(
    json_output,
    orient="records",
    date_format="iso",
    indent=2
)

print(f"JSON berhasil disimpan: {json_output}")
print(f"Ukuran file: {json_output.stat().st_size:,} bytes")
```

```python
# ============================================================
# NO. 5 BAGIAN 6 - LOAD KE PARQUET
# ============================================================

parquet_output = output_path / "sensor_data_clean.parquet"

df_clean.to_parquet(
    parquet_output,
    engine="fastparquet",
    index=False
)

print(f"Parquet berhasil disimpan: {parquet_output}")
print(f"Ukuran file: {parquet_output.stat().st_size:,} bytes")
```

```python
# ============================================================
# NO. 5 BAGIAN 7 - PERBANDINGAN UKURAN FILE
# ============================================================

file_sizes = pd.DataFrame({
    "Format": [
        "CSV",
        "Excel",
        "JSON",
        "Parquet"
    ],
    "File Size (bytes)": [
        csv_file.stat().st_size,
        excel_output.stat().st_size,
        json_output.stat().st_size,
        parquet_output.stat().st_size
    ]
})

file_sizes["File Size (KB)"] = (
    file_sizes["File Size (bytes)"] / 1024
).round(2)

display(file_sizes)
```
